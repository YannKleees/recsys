```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (16, 6)

# Load the processed core data
try:
    df = pd.read_parquet("HotelRec_core.parquet")
    print(f"Successfully loaded 'HotelRec_core.parquet'. Shape: {df.shape}")
    print("\nData Schema (dtypes):")
    print(df.info())
except FileNotFoundError:
    print("Error: 'HotelRec_core.parquet' not found. Please run the preprocessing script first.")
    exit()

# 1. Calculate reviews per author (user)
author_counts = df['author_id'].value_counts()

# 2. Calculate reviews per hotel
hotel_counts = df['hotel_id'].value_counts()

# --- Visualization ---

# Create a figure with two subplots
fig, (ax1, ax2) = plt.subplots(1, 2)
fig.suptitle('Distribution of Reviews per User and per Hotel', fontsize=16)

# Plot for reviews per user
sns.histplot(author_counts, bins=50, ax=ax1, log_scale=(False, True))
ax1.set_title('Reviews per User (Author)')
ax1.set_xlabel('Number of Reviews')
ax1.set_ylabel('Number of Users (log scale)')

# Plot for reviews per hotel
sns.histplot(hotel_counts, bins=50, ax=ax2, log_scale=(False, True))
ax2.set_title('Reviews per Hotel')
ax2.set_xlabel('Number of Reviews')
ax2.set_ylabel('Number of Hotels (log scale)')

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()

# --- Print Descriptive Statistics ---
print("\n--- Sparsity Analysis ---")
print("\nStatistics for Reviews per User:")
print(author_counts.describe())

print("\nStatistics for Reviews per Hotel:")
print(hotel_counts.describe())
```
